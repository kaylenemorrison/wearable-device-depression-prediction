```python
import pandas as pd
import numpy as np
```

```python
PATH = 'data_cleaned/final_dataset.csv'
```

```python
df = pd.read_csv(PATH)
```

```python
print('\nShape:', df.shape)
```

```python
print('\nSample:')
print(df.head(3))
```

```python
print('\nColumn dtypes:')
print(df.dtypes)
```

```python
# Missingness summary
missing_count = df.isna().sum()
missing_pct = (df.isna().mean() * 100)
missing_summary = pd.concat([missing_count, missing_pct], axis=1)
missing_summary.columns = ['missing_count', 'missing_pct']
missing_summary = missing_summary.sort_values('missing_count', ascending=False)

print('\nTop columns by missingness:')
print(missing_summary)
```

```python
# Numeric summary
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"\nNumeric columns ({len(num_cols)}):", num_cols)
if num_cols:
    print(df[num_cols].describe().transpose())
```
