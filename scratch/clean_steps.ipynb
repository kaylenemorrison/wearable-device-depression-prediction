```python
import pandas as pd
```

```python
steps_raw = pd.read_csv("../data/radarV1_MDD/csv_files/dailyFeatures_fitbit_steps.csv")
```

```python
steps_raw.columns
```

```python
steps_raw["Script_Version"].unique()
```

```python
steps_v1 = steps_raw[steps_raw["Script_Version"] == "v1"]
```

```python
duplicates_mask_v1 = steps_v1.duplicated(subset=["SUBJECT_ID", "DATE_ID"], keep=False)
```

```python
duplicates_mask_v1.sum()
```

```python
steps_v1_missing = steps_v1.isnull().mean().sort_values(ascending=False)
```

```python
for col, val in steps_v1_missing.items():
    print(f"  - {col}: {val:.2f}% missing")
```

```python
steps_v1_cleaned = steps_v1.dropna(how='all', axis=1) # subset columns that are completely missing data
```

```python
steps_v1_missing = steps_v1_cleaned.isnull().mean().sort_values(ascending=False)
```

```python
for col, val in steps_v1_missing.items():
    print(f"  - {col}: {val:.2f}% missing")
```
