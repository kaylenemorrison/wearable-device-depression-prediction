```python
# 1. Convert the string column to datetime objects
sleep_data['updated_at'] = pd.to_datetime(sleep_data['updated_at'])

# 2. Extract just the date (removing the time component)
sleep_data['date'] = sleep_data['updated_at'].dt.date

# 3. Identify duplicates based on ID and the new date column
duplicates = sleep_data[sleep_data.duplicated(subset=['participant_name', 'date'], keep=False)]

# 4. Get a list of unique IDs that have at least one duplicate date
repeat_participants = duplicates['participant_name'].unique()

print(f"Found {len(repeat_participants)} participants with duplicate dates.")
```

```python
def count_participant_days(df, date_column):
    """
    Counts unique dates per participant and returns a sorted Series.
    """
    # Ensure date column is present; if not, create it from updated_at
    if 'date' not in df.columns:
        df['date'] = pd.to_datetime(df[f"{date_column}"]).dt.date
    
    # Group by participant and count unique dates
    day_counts = df.groupby('participant_name')['date'].nunique().sort_values(ascending=False)
    
    return day_counts

```

```python
# Usage:
participant_engagement = count_participant_days(sleep_data, 'updated_at')
print(participant_engagement)
```

```python
# 1. Convert the string column to datetime objects
phq8_data['created_at'] = pd.to_datetime(phq8_data['created_at'])

# 2. Extract just the date (removing the time component)
phq8_data['date'] = phq8_data['created_at'].dt.date

# 3. Identify duplicates based on ID and the new date column
duplicates = phq8_data[phq8_data.duplicated(subset=['participant_name', 'date'], keep=False)]

# 4. Get a list of unique IDs that have at least one duplicate date
repeat_participants = duplicates['participant_name'].unique()

print(f"Found {len(repeat_participants)} participants with duplicate dates.")
```

```python
# 1. Convert Unix timestamps to datetime
# unit='s' is used for timestamps in seconds (10 digits)
phq8_data['file_modified_on_datetime'] = pd.to_datetime(phq8_data['file_modified_on'], unit='s')

# 2. Extract just the date (YYYY-MM-DD)
phq8_data['date'] = phq8_data['file_modified_on_datetime'].dt.date

# 3. Identify duplicates based on ID and the extracted date
duplicates = phq8_data[phq8_data.duplicated(subset=['participant_name', 'date'], keep=False)]

# 4. Get the unique IDs of participants with repeating dates
repeat_participants = duplicates['participant_name'].unique()

print(f"Found {len(repeat_participants)} participants with duplicate dates.")
```

```python
phq8_data['time_completed_datetime'] = pd.to_datetime(phq8_data['time_completed_local'])

# 2. Extract just the date (YYYY-MM-DD)
phq8_data['date'] = phq8_data['time_completed_datetime'].dt.date

# 3. Identify duplicates based on ID and the extracted date
duplicates = phq8_data[phq8_data.duplicated(subset=['participant_name', 'date'], keep=False)]

# 4. Get the unique IDs of participants with repeating dates
repeat_participants = duplicates['participant_name'].unique()

print(f"Found {len(repeat_participants)} participants with duplicate dates.")
```

```python
phq8_data['end_time7_datetime'] = pd.to_datetime(phq8_data['end_time7'], unit='s')

# 2. Extract just the date (YYYY-MM-DD)
phq8_data['date'] = phq8_data['end_time7_datetime'].dt.date

# 3. Identify duplicates based on ID and the extracted date
duplicates = phq8_data[phq8_data.duplicated(subset=['participant_name', 'date'], keep=False)]

# 4. Get the unique IDs of participants with repeating dates
repeat_participants = duplicates['participant_name'].unique()

print(f"Found {len(repeat_participants)} participants with duplicate dates.")
```
